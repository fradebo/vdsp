```python
def swap_qubit(circuit, pos_start):
    c.add((pos_start+1, pos_start+2), symb.BS.H(np.pi))
    c.add((pos_start, pos_start+1), symb.BS.H(np.pi))
    c.add((pos_start+2, pos_start+3), symb.BS.H(np.pi))
    c.add((pos_start+1, pos_start+2), symb.BS.H(np.pi))

c = pcvl.Circuit(19*6+2)
fuse2(c,Qbit(0),Qbit(2)) #0,12
swap_qubit(c,6)
fuse2(c,Qbit(4),Qbit(6)) #0,8
swap_qubit(c,12)
fuse2(c,Qbit(10),Qbit(12)) #0,2
swap_qubit(c,18)
fuse2(c,Qbit(16),Qbit(18)) #0,5
swap_qubit(c,24)
fuse2(c,Qbit(22),Qbit(24)) #0,1

swap_qubit(c,26)
swap_qubit(c,30)
fuse2(c,Qbit(28),Qbit(30)) #5,15
swap_qubit(c,32)
swap_qubit(c,34)
swap_qubit(c,36)

swap_qubit(c,34)
fuse2(c,Qbit(32),Qbit(34)) #5,10

swap_qubit(c,36)
swap_qubit(c,38)
swap_qubit(c,40)
fuse2(c,Qbit(42),Qbit(44)) #1,14

swap_qubit(c,48)
fuse2(c,Qbit(46),Qbit(48)) #1,6
swap_qubit(c,54)
fuse2(c,Qbit(52),Qbit(54)) #1,4
swap_qubit(c,60)
fuse2(c,Qbit(58),Qbit(60)) #1,3
swap_qubit(c,66)
fuse2(c,Qbit(64),Qbit(66)) #1,7

for i in range(72,96,2):
    swap_qubit(c,i) #swap 7 to 18

swap_qubit(c,98)
fuse2(c,Qbit(96),Qbit(98)) #7,18
swap_qubit(c,104)
fuse2(c,Qbit(102),Qbit(104)) #7,17
swap_qubit(c,110)
fuse2(c,Qbit(108),Qbit(110)) #7,13
#116 modes in total? 

swap_qubit(c,68)
fuse2(c,Qbit(70),Qbit(72)) #3,11

for i in range(62,76,2):
    swap_qubit(c,i) #swap 4 to 16 

fuse2(c,Qbit(76),Qbit(78)) #4,16
fuse2(c,Qbit(82),Qbit(84)) #16,19

for i in range(56,88,2):
    swap_qubit(c,i) #swap 6 to 9 

fuse2(c,Qbit(88),Qbit(90)) #4,16

pcvl.pdisplay(c).save_svg('largetreeghz.svg')
```
