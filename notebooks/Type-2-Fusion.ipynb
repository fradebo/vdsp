```python
import sys
sys.path.append('..')
import perceval as pcvl
import perceval.components as symb
from perceval.utils import StateGenerator, Encoding
import numpy as np
import networkx as nx
```

```python
G = nx.Graph([[0,1],[2,3]])
bsg = pcvl.Circuit(8, name = "Fusion50")
bsg.add((0,1), symb.BS.H())
bsg.add((2,3), symb.BS.H())
bsg.add((1,2), symb.BS.H(np.pi))
bsg.add((0,1), symb.BS.H())
bsg.add((2,3), symb.BS.H())

# Set the generator with the dual rail encoding
generator=StateGenerator(Encoding.DUAL_RAIL)
instate = generator.graph_state(G)
print(instate)
```

```python
backend = pcvl.backends.NaiveBackend()
backend.set_circuit(bsg)

sim = pcvl.simulators.Simulator(backend)
```

```python
sim.evolve(instate)
```
