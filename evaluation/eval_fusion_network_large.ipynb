```python
def connect_graph(G):
    C = list(nx.connected_components(G))
    for idx in range(1,len(C)):
        G.add_edge(list(C[idx-1])[0],list(C[idx])[0])

# random.seed(1234)   
random.seed(4325) 
success_probs = [0.5,0.66,0.75,0.85,1]
# names = [('dfs', 'random', True), ('dfs', 'random', False),('dfs', 'min', True),('dfs', 'min', False),('dfs', 'max', True),('dfs', 'max', False),('bfs', 'random', True),('bfs', 'random', False),('bfs', 'min', True),('bfs', 'min', False),('bfs', 'max', True),('bfs', 'max', False)]
names = [('dfs', 'random'),('dfs', 'min'),('dfs', 'max'),('bfs', 'random'),('bfs', 'min'),('bfs', 'max'),('rand')]
strategy_improvement = {name: {p: [] for p in success_probs} for name in names}

for i in range(10):
    G = nx.erdos_renyi_graph(8, 0.3)
    connect_graph(G)
    nx.draw(G, with_labels=True)
    print(G.edges)
    plt.show()
    fpt_bins = {p: [] for p in success_probs}
    for conf in itertools.product(['dfs','bfs'],['random','min','max']): #
        T = get_spanning_tree(G,conf[0],conf[1],False)
        max_deg_vertex = sorted([(G.degree(v),v) for v in G.nodes])[-1][1]
        fusion_network = build_fusion_network_from_spanning_tree(T,max_deg_vertex,G)
        fusion_order = approximate_minimal_fpt_fusion_order(fusion_network)
        print("tree fusion_order",fusion_order)
        max_qubit = int(max([q for f in fusion_order for q in f])/3)*3+3
        fpt = get_fp_times(fusion_order, max_qubit, success_probs)
#         print("fpt",fpt)
        for k,v in fpt.items():
            fpt_bins[k].append(v)
    
    #random
    fusion_network_rand = build_fusion_network_random(G)
    fusion_order = approximate_minimal_fpt_fusion_order(fusion_network_rand)
    if fusion_order[0] == (5, 6):
        import pdb
        pdb.set_trace()
    print("random fusion_order",fusion_order)
    max_qubit = int(max([q for f in fusion_order for q in f])/3)*3+3
    fpt = get_fp_times(fusion_order, max_qubit, success_probs)
    print("random fpt",fpt)
    for k,v in fpt.items():
        fpt_bins[k].append(v)
    #ghz
#     ghz_partitions = ghz_partitioning_heuristic(G)
#     fusion_network_ghz = build_fusion_network_from_ghz_partitions(ghz_partitions,G)
#     fusion_order = approximate_minimal_fpt_fusion_order(fusion_network_ghz)
#     print("ghz fusion_order",fusion_order)
#     max_qubit = int(max([q for f in fusion_order for q in f])/3)*3+3
#     fpt = get_fp_times(fusion_order, max_qubit, success_probs)
# #     print('fpt ghz:',fpt)
#     for k,v in fpt.items():
#         fpt_bins[k].append(v)
    
#     import pdb
    
#     pdb.set_trace()
    for k,v in fpt_bins.items():
#         print("succes prob",k)
        avg = sum(v)/len(v)
        for i in range(len(v)):
            strategy_improvement[names[i]][k].append(1-v[i]/avg)
#             print("strategy",names[i],"improve",1-v[i]/avg)
#         print(conf,":",fpt)
```

```python
avg_strategy_improvement = {name: {p: sum(strategy_improvement[name][p])/len(strategy_improvement[name][p]) for p in success_probs} for name in names}
avg_strategy_improvement
```

```python
import matplotlib.pyplot as plt
import matplotlib

def plot_data(data):
    # Get all unique string labels
    labels = list(data.keys())

    # Initialize a list to store the colors for each label
    colors = matplotlib.cm.tab20(range(len(labels)))


    # Iterate over the labels and their corresponding data
    for i, (label, value) in enumerate(data.items()):
        # Get the probabilities and average values
        probs = list(value.keys())
        avgs = list(value.values())

        # Plot the data
        plt.plot(probs, avgs, label=label, color=colors[i])

    # Set the title and labels
    plt.title('Average Values vs Probabilities')
    plt.xlabel('Probabilities')
    plt.ylabel('Average Values')

    # Add a legend
    plt.legend(loc='lower center', bbox_to_anchor=(0.5, -0.4), ncol=len(labels)/3)
    plt.savefig('random83.png', bbox_inches='tight')
    # Show the plot
    plt.show()

plot_data(avg_strategy_improvement)
```
